```python

import os
import sys
from pathlib import Path
from dotenv import load_dotenv

# Cargar las variables de entorno desde el archivo .env
load_dotenv()

# Obtener la ruta desde las variables de entorno
scripts_path = Path(os.getenv("SCRIPTS_PATH"))
sys.path.append(str(scripts_path))

from config import DB_HOST, DB_USER, DB_PASSWORD, DB_NAME  # Importar credenciales
from queries import QUERY_qx  # Importar la consulta SQL
```

```python
import pandas as pd
import mariadb

# Establecer conexión con la base de datos
connection = mariadb.connect(
    host=DB_HOST,
    user=DB_USER,
    password=DB_PASSWORD,
    database=DB_NAME
)

# Crear un cursor y ejecutar la consulta
cursor = connection.cursor()
cursor.execute(QUERY_qx)

# Obtener los resultados del query
result = cursor.fetchall()

# Convertir el resultado del query en un DataFrame
columns = [desc[0] for desc in cursor.description]  # Obtener los nombres de las columnas
df_result = pd.DataFrame(result, columns=columns)

# Cerrar el cursor y la conexión
cursor.close()
connection.close()

# Cargar la ruta de salida desde las variables de entorno
output_path = Path(os.getenv("DATA_PATH")) / "resultado_QUERY_qx.csv"

# Guardar el DataFrame como un archivo CSV
df_result.to_csv(output_path, index=False, encoding="utf-8")

print(f"El archivo CSV se ha guardado en: {output_path}")
```

```python
# Renombrar las columnas del DataFrame
df_result.rename(columns={
    'BF5BBFI1221BI_B2129FIJF': 'Descripción',
    'BF5BBFI1221BI_F5997JB9F': 'Tipo',
    'BF5BBFI1221BI_Y7FY7YF1B': 'Clave',
    '-- Clave\nCOUNT(BF5BBFI1221BI_Y7FY7YF1B)': 'Conteo por Clave'
}, inplace=True)

# Mostrar el DataFrame con los nuevos nombres de columnas
df_result
```

```python
# Convertir la columna 'Tipo' a tipo categórico
df_result['Tipo'] = df_result['Tipo'].astype('category')

# Verificar el cambio de tipo
df_result.dtypes
```
